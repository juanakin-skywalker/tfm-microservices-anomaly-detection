## <span style="color:#2ca02c">2.1 IF con vectores que usan datos de 3 puntos seguidos con la misma etiqueta</span>

```python
def X_matrix_sel_data(X_matrix,lista_feature_names,lista_idx_filas=None,lista_feature_sel=None):
    #X_matrix es la matriz completa
    #lista_feature_names son los nombres de las columnas
    #lista_idx_filas es la lista de indices de las filas que tomaremos
    #lista_feature_sel es la lista de nombres de las columnas que tomaremos, es un subconjunto de lista_feature_names
    if lista_idx_filas is None:
        lista_idx_filas   = range(0,X_matrix.shape[0])
    if lista_feature_sel is None:
        lista_feature_sel = lista_feature_names.copy()

    lista_feature_idx = []
    idx=0
    for feature in lista_feature_names:
        if feature in lista_feature_sel:
            lista_feature_idx.append(idx)
        idx += 1        

    # 1. Filtramos primero las filas
    X_temp = X_matrix[list(lista_idx_filas), :]
    
    # 2. Filtramos después las columnas sobre el resultado anterior
    X_matrix_sub = X_temp[:, lista_feature_idx]
    
    return X_matrix_sub
```

```python
idx_correct_train = dic_idx_train['correct']
idx_correct_val   = dic_idx_val['correct']
idx_correct_test  = dic_idx_test['correct']
```

```python
N_datos_test_por_anomalia = int( len(idx_correct_test)  / (len(lista_label)-1) )
```

```python
idx_test  = idx_correct_test.copy()
idx_val   = idx_correct_val.copy()

for label in lista_label:
    if label=='correct':
        continue
    if len(dic_idx_test[label])<N_datos_test_por_anomalia:

        idx_test = np.concatenate((idx_test , dic_idx_test[label]))
        idx_val  = np.concatenate((idx_val  , dic_idx_val[label]))
    else:
        #idx_test.extend(dic_idx_test[label][:N_datos_test_por_anomalia])
        # Corrección: pon los dos arrays dentro de un paréntesis

        idx_test = np.concatenate((idx_test   , dic_idx_test[label][:N_datos_test_por_anomalia]))
        idx_val  = np.concatenate((idx_val    , dic_idx_val[label][:N_datos_test_por_anomalia]))


X_val   = X_matrix[idx_val]
X_test  = X_matrix[idx_test]
```

```python
#conjunto de datos de entrenamiento con la mitad de datos correctos y la mitad anomalias
idx_train = idx_correct_train.copy()
N_datos_train_por_anomalia = int( len(idx_correct_train)  / (len(lista_label)-1) )
for label in lista_label:
    if label=='correct':
        continue
    if len(dic_idx_train[label])<N_datos_test_por_anomalia:
        idx_train = np.concatenate((idx_train , dic_idx_train[label]))

    else:
        #idx_test.extend(dic_idx_test[label][:N_datos_test_por_anomalia])
        # Corrección: pon los dos arrays dentro de un paréntesis
        idx_train = np.concatenate((idx_train , dic_idx_train[label][:N_datos_train_por_anomalia]))

X_train_clasificacion  = X_matrix[idx_train]


```

```python
Y_train_clasificacion = df_result.loc[idx_train, 'Y'].to_numpy()
```

```python
X_correct_train = X_matrix[idx_correct_train]
X_correct_test  = X_matrix[idx_correct_test]
```

```python
df_result = pd.DataFrame({
    't_rel': df.loc[idx_test, 't_rel'],
    'execution_name': df.loc[idx_correct_test, 'execution_name'],
    'label': df.loc[idx_test, 'label']
})

N = 0
lista_result=[]
for contaminacion in [0.02, 0.015, 0.01, 0.008, 0.006, 0.004, 0.0035, 0.003, 0.002, 0.001, 0.0005]:
    N += 1
    model = IsolationForest(
        n_estimators=200, 
        max_samples='auto', 
        contamination=contaminacion, 
        random_state=SEED
    )
    model.fit(X_correct_train)
    scores = model.decision_function(X_test)
    Y_pred = model.predict(X_test)

    model_name = f'IF(contaminacion={contaminacion})_3puntos'

    
    df_result[f'result_({model_name})'] = float('nan')
    df_result[f'scores_({model_name})'] = float('nan')
    df_result.loc[idx_test, f'result_({model_name})'] = Y_pred
    df_result.loc[idx_test, f'scores_({model_name})'] = scores
    
    # Mapeo: 1 para Anomalía (Positivo) y 0 para Normal (Negativo)
    y_true = df_result.loc[idx_test, 'label'].apply(lambda x: 0 if x == 'correct' else 1)
    y_pred = df_result.loc[idx_test, f'result_({model_name})'].apply(lambda x: 0 if x == 1 else 1)
    
    # Invertimos los scores para que valores más altos correspondan a la anomalía
    y_scores_roc = -scores 
    roc_auc = roc_auc_score(y_true, y_scores_roc)

    # 1. Componentes de la matriz de confusión
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    
    # 2. Cálculo de métricas
    acc = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    f1 = f1_score(y_true, y_pred, zero_division=0)

    parametros = {'contaminacion': contaminacion}
    
    # 3. Guardar en un diccionario de registro
    reg_resultado_algoritmo = {
        'test_name': model_name,
        'parametros': json.dumps(parametros),
        'TP': int(tp),
        'FP': int(fp),
        'TN': int(tn),
        'FN': int(fn),
        'accuracy': float(acc),
        'precision': float(precision),
        'recall': float(recall),
        'specificity': float(specificity),
        'f1_score': float(f1),
        'roc_auc': float(roc_auc)
    }
    lista_result.append(reg_resultado_algoritmo)
    
    # 4. Visualización conjunta (Matriz de Confusión a la izquierda, Curva ROC a la derecha)
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    # Gráfica izquierda: Matriz de Confusión
    disp_cm = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal (0)', 'Anomalía (1)'])
    disp_cm.plot(ax=axes[0], cmap='Blues', colorbar=False)
    axes[0].set_title(f'Matriz de Confusión\n{model_name}')
    
    # Gráfica derecha: Curva ROC
    RocCurveDisplay.from_predictions(
        y_true, 
        y_scores_roc, 
        ax=axes[1], 
        name=f'IF (AUC = {roc_auc:.3f})'
    )
    axes[1].plot([0, 1], [0, 1], 'k--', label='Aleatorio (AUC = 0.5)')
    axes[1].set_title(f'Curva ROC\n{model_name}')
    axes[1].legend(loc='lower right')
    
    plt.tight_layout()
    plt.show()
    paquete = {
        'model_name' : model_name,
        'model'      : model,
        'parametros' : {'contaminacion':contaminacion},
        'resultado'  : {'accuracy':acc,'precision':precision,'recall':recall,'f1':f1,'specificity':specificity}
    }
    joblib.dump(paquete, f'{FOLDER_SALIDA}/models/{model_name}.joblib')
    
# DataFrame global final
df_anomaly_detection_summary = pd.DataFrame(lista_result)
```

```python
def custom_f1_scorer(estimator, X, y_true):
    # Obtenemos las predicciones del modelo (-1 y 1)
    y_pred = estimator.predict(X)
    
    # INVERTIMOS EL MAPEO: 
    # Si el modelo predice 1 (normal para sklearn), lo pasamos a 0 (normal para tu Y_test)
    # Si el modelo predice -1 (anomalía para sklearn), lo pasamos a 1 (anomalía para tu Y_test)
    y_pred_binary = np.where(y_pred == 1, 0, 1)
    
    y_true_binary = np.asarray(y_true).astype(int)
    
    return f1_score(y_true_binary, y_pred_binary, pos_label=1, average='binary', zero_division=0)
    
# Ejecutamos la importancia de permutación con el scorer corregido
result = permutation_importance(
    model, X_test, Y_test,
    scoring=custom_f1_scorer,
    n_repeats=20,
    n_jobs=-1,
    random_state=SEED
)


```

```python
N_puntos=3
# 1. Definir o recuperar las características base iniciales y la lista expandida con lags
features_base_actuales = sorted(list(set([f.split('(lag_')[0] for f in lista_metric_name_vector])))
def expandir_con_lags(features_base_list, n_puntos):
    lista_expandida = []
    for feature in features_base_list:
        if n_puntos > 1:
            for lag in range(0, n_puntos):
                lista_expandida.append(f'{feature}(lag_{lag})')
        else:
            lista_expandida.append(feature)
    return lista_expandida

lista_features_actual = expandir_con_lags(features_base_actuales, N_puntos)

# 2. Asignar tus datos de entrenamiento
x_train_xgb = X_train_clasificacion
y_train_xgb = Y_train_clasificacion

# 3. Entrenar XGBoost
xgb_model = xgb.XGBClassifier(
    n_estimators=150,
    max_depth=6,
    learning_rate=0.05,
    random_state=SEED,
    n_jobs=-1
)
xgb_model.fit(x_train_xgb, y_train_xgb)

# 4. Obtener importancias y agregarlas sumando los lags de cada variable base
importancias_brutas = xgb_model.feature_importances_
dict_importancias = {feat: imp for feat, imp in zip(lista_features_actual, importancias_brutas)}

importancias_base = []
for base in features_base_actuales:
    importancia_total_lags = sum(dict_importancias.get(f'{base}(lag_{lag})', 0) for lag in range(N_puntos))
    importancias_base.append({'caracteristica': base, 'importancia': importancia_total_lags})

df_imp_xgb = pd.DataFrame(importancias_base)

# 5. Ordenar las características base de menor a mayor importancia según XGBoost
features_base_ordenadas = df_imp_xgb.sort_values(by='importancia', ascending=True)['caracteristica'].tolist()




```

```python


# 1. Definir función de evaluación completa con el modelo Isolation Forest entrenado solo con datos limpios
def evaluar_features_if(features_base_list):
    lista_feat_exp = expandir_con_lags(features_base_list, N_puntos)
    
    X_train_sub = X_matrix_sel_data(X_matrix, lista_metric_name_vector, idx_correct_train, lista_feat_exp)
    X_test_sub  = X_matrix_sel_data(X_matrix, lista_metric_name_vector, idx_test, lista_feat_exp)
    
    model_if = IsolationForest(
        n_estimators=200, 
        max_samples='auto', 
        contamination=0.001, 
        random_state=SEED
    )
    model_if.fit(X_train_sub)
    
    scores = model_if.decision_function(X_test_sub)
    Y_pred = model_if.predict(X_test_sub)
    y_pred_bin = np.where(Y_pred == 1, 0, 1)
    
    y_true_test = df_completo.loc[idx_test, 'Y'].to_numpy()
    
    cm = confusion_matrix(y_true_test, y_pred_bin, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    
    acc = accuracy_score(y_true_test, y_pred_bin)
    precision = precision_score(y_true_test, y_pred_bin, zero_division=0)
    recall = recall_score(y_true_test, y_pred_bin, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    f1 = f1_score(y_true_test, y_pred_bin, pos_label=1, average='binary', zero_division=0)
    roc_auc = roc_auc_score(y_true_test, -scores)
    
    metricas = {
        'N_feat_base': len(features_base_list),
        'N_feat_total': len(lista_feat_exp),
        'accuracy': float(acc),
        'precision': float(precision),
        'recall': float(recall),
        'specificity': float(specificity),
        'f1': float(f1),
        'roc_auc': float(roc_auc)
    }
    
    return metricas, model_if

# 2. Inicializar lista para almacenar el historial de resultados
historial_resultados_if = []

# Partimos de la lista completa ordenada por XGBoost (de menor a mayor importancia)
features_actuales = list(features_base_ordenadas)

# 3. Bucle de reducción: evaluamos y vamos quitando la menos importante (primera de la lista) hasta llegar a 10 features base
while len(features_actuales) >= 1:
    metricas_ciclo, modelo_ciclo = evaluar_features_if(features_actuales)
    
    historial_resultados_if.append({
        **metricas_ciclo,
        'features_base': list(features_actuales),
        'features_totales': list(expandir_con_lags(features_actuales, N_puntos)),
        'modelo': modelo_ciclo
    })
    
    print(f"N_feat_base: {metricas_ciclo['N_feat_base']:3d} | N_total: {metricas_ciclo['N_feat_total']:3d} | "
          f"Accuracy: {metricas_ciclo['accuracy']:.4f} | Precision: {metricas_ciclo['precision']:.4f} | "
          f"Specificity: {metricas_ciclo['specificity']:.4f} | Recall: {metricas_ciclo['recall']:.4f} | "
          f"F1: {metricas_ciclo['f1']:.4f} | AUC: {metricas_ciclo['roc_auc']:.4f}")
    
    # Si ya estamos en 10, rompemos antes de vaciar la lista
    if len(features_actuales) == 1:
        break
        
    # Eliminamos la característica base menos importante para el siguiente ciclo
    features_actuales.pop(0)

print("\nBarrido descendente completado. La lista 'historial_resultados_if' contiene todos los indicadores y modelos entrenados.")
```

## <span style="color:#2ca02c">2.2 IF con vectores que usan datos de 1 puntos seguidos con la misma etiqueta</span>

```python
# 1. Calculamos cuántas columnas representa el primer tercio
num_columnas_total = X_matrix.shape[1]
primer_tercio_n = num_columnas_total // 3  # División entera

print(f"Total de columnas: {num_columnas_total}")
print(f"Nos quedamos con el primer tercio: {primer_tercio_n} columnas")

# 2. Seleccionamos el primer tercio de las columnas en la matriz global
# (Asegúrate de aplicar esto antes de tus particiones de train/test)
X_matrix_reducida = X_matrix[:, :primer_tercio_n]

# 3. Recortamos los datos de entrenamiento y test con este nuevo subconjunto
X_correct_train = X_matrix_reducida[idx_correct_train]
X_correct_test  = X_matrix_reducida[idx_correct_test]
X_test          = X_matrix_reducida[idx_test]
X_val           = X_matrix_reducida[idx_val]

# 4. Ajustamos también el vector de nombres de las características para que coincida exactamente
lista_metric_name_vector = lista_metric_name_vector[:primer_tercio_n]
```

```python
N = 0
lista_result = []

# 1. Definimos y_true FUERA del bucle (o al inicio absoluto) ya que las etiquetas de test no cambian nunca
y_true = df_result.loc[idx_test, 'label'].apply(lambda x: 0 if x == 'correct' else 1)

for contaminacion in [0.02, 0.015, 0.01, 0.008, 0.006, 0.004, 0.0035, 0.003, 0.002, 0.001, 0.0005]:
    model_name = f'IF(contaminacion={contaminacion})_1punto'
    N += 1
    
    # 2. Entrenar el Isolation Forest
    model = IsolationForest(
        n_estimators=200, 
        max_samples='auto', 
        contamination=contaminacion,        
        random_state=SEED
    )
    model.fit(X_correct_train)
    scores = model.decision_function(X_test)
    
    # 3. Preparar scores para ROC
    y_scores_roc = -scores  
    roc_auc = roc_auc_score(y_true, y_scores_roc)
    
    # 4. Cálculo del umbral óptimo mediante el índice de Youden en cada iteración
    fpr, tpr, thresholds = roc_curve(y_true, y_scores_roc)
    optimal_idx = np.argmax(tpr - fpr)
    optimal_threshold = thresholds[optimal_idx]
    
    # 5. Aplicar el umbral de Youden real
    #y_pred = (y_scores_roc >= optimal_threshold).astype(int)
    y_pred_raw = model.predict(X_test)
    y_pred = np.where(y_pred_raw == -1, 1, 0)

    
    # Guardar en el DataFrame global
    df_result[f'result_({model_name})'] = float('nan')
    df_result[f'scores_({model_name})'] = float('nan')
    df_result.loc[idx_test, f'result_({model_name})'] = y_pred
    df_result.loc[idx_test, f'scores_({model_name})'] = scores
    
    # 6. Métricas y matriz de confusión
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    
    acc = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    f1 = f1_score(y_true, y_pred, zero_division=0)
    

    print(f'Accuracy    = {acc}')
    print(f'Precision   = {precision}')
    print(f'Recall      = {recall}')
    print(f'Specificity = {specificity}')    
    print(f'F1          = {f1}')
    
    parametros = {'contaminacion': contaminacion, 'metodo': 'youden', 'umbral_youden': float(optimal_threshold)}
    
    reg_resultado_algoritmo = {
        'test_name': model_name,
        'parametros': json.dumps(parametros),
        'TP': int(tp), 'FP': int(fp), 'TN': int(tn), 'FN': int(fn),
        'accuracy': float(acc),
        'precision': float(precision),
        'recall': float(recall),
        'specificity': float(specificity),
        'f1_score': float(f1),
        'roc_auc': float(roc_auc)
    }
    lista_result.append(reg_resultado_algoritmo)
    
    # 7. Visualización
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    disp_cm = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal (0)', 'Anomalía (1)'])
    disp_cm.plot(ax=axes[0], cmap='Blues', colorbar=False)
    axes[0].set_title(f'Matriz de Confusión \n{model_name}')
    
    RocCurveDisplay.from_predictions(y_true, y_scores_roc, ax=axes[1], name=f'IF (AUC = {roc_auc:.3f})')
    axes[1].plot([0, 1], [0, 1], 'k--', label='Aleatorio (AUC = 0.5)')
    axes[1].set_title(f'Curva ROC\n{model_name}')
    axes[1].legend(loc='lower right')
    
    plt.tight_layout()
    plt.show()


    paquete = {
        'model_name' : model_name,
        'model'      : model,
        'parametros' : {'contaminacion':contaminacion},
        'resultado'  : {'accuracy':acc,'precision':precision,'recall':recall,'f1':f1,'specificity':specificity}
    }
    joblib.dump(paquete, f'{FOLDER_SALIDA}/models/{model_name}.joblib')

df_anomaly_detection_summary = pd.DataFrame(lista_result)
```

# <span style="color:#2ca02c">3. Usamos Isolation Forest (IF) clasificar anomalias</span>

<span style="color:#2ca02c">Crearemos un modelo IF entrenado con datos de anomalia, para clasificar primero se aplciara el detector de anomalia y luego aplica los detectores individuales para cada anomalia</span>

```python
from sklearn.metrics import roc_auc_score, RocCurveDisplay

lista_result=[]
contaminacion=0.001
dic_model={}
for label in lista_label:
    if label=='correct':
        continue
    idx_train = dic_idx_train[label]
    X_train   = X_matrix[idx_train]
 
    dic_model[label] = IsolationForest(
        n_estimators=200, 
        max_samples='auto', 
        contamination=contaminacion, 
        random_state=SEED
    )
    N_datos_test_por_anomalia = int( len(dic_idx_test[label])  / (len(lista_label)-2) )+1
    idx_test = np.array([], dtype=int)
    idx_test_tmp = np.array([], dtype=int)
    for label2 in lista_label:
        if label2=='correct':
            continue
        elif label2==label:
            idx_test_tmp=dic_idx_test[label2]
        else:
            idx_test_tmp=dic_idx_test[label2][:N_datos_test_por_anomalia]
            
        idx_test=np.concatenate((idx_test, idx_test_tmp))
        
    X_test   = X_matrix[idx_test]
    dic_model[label].fit(X_train)
    Y_pred = dic_model[label].predict(X_test)
    scores = dic_model[label].decision_function(X_test)
    threshold = np.percentile(scores, 20)
    y_pred = np.where(scores < threshold, 1, 0)

    y_true = df.loc[idx_test, 'label'].apply(lambda x: 0 if x == label else 1)
    
    # Invertimos los scores para el cálculo del ROC-AUC (valores mayores = mayor probabilidad de anomalía)
    y_scores_roc = -scores 
    roc_auc = roc_auc_score(y_true, y_scores_roc)

    # 1. Componentes de la matriz de confusión
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    
    # 2. Cálculo de métricas
    acc = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)         # Sensibilidad / True Positive Rate
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0          # True Negative Rate
    f1 = f1_score(y_true, y_pred, zero_division=0)

    model_name=f'IF(label={label})'
    parametros={'contaminacion':contaminacion}
    
    # 3. Guardar en un diccionario de registro (añadiendo roc_auc)
    reg_resultado_algoritmo = {
        'test_name': model_name,
        'parametros': json.dumps(parametros),
        'TP': int(tp),
        'FP': int(fp),
        'TN': int(tn),
        'FN': int(fn),
        'accuracy': float(acc),
        'precision': float(precision),
        'recall': float(recall),
        'specificity': float(specificity),
        'f1_score': float(f1),
        'roc_auc': float(roc_auc)
    }
    lista_result.append(reg_resultado_algoritmo)
    print(reg_resultado_algoritmo)
    
    # 4. Visualización conjunta (Matriz de Confusión a la izquierda, Curva ROC a la derecha)
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    # Gráfica izquierda: Matriz de Confusión
    disp_cm = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal (0)', 'Anomalía (1)'])
    disp_cm.plot(ax=axes[0], cmap='Blues', colorbar=False)
    axes[0].set_title(f'Matriz de Confusión\n{model_name}')
    
    # Gráfica derecha: Curva ROC
    RocCurveDisplay.from_predictions(
        y_true, 
        y_scores_roc, 
        ax=axes[1], 
        name=f'IF (AUC = {roc_auc:.3f})'
    )
    axes[1].plot([0, 1], [0, 1], 'k--', label='Aleatorio (AUC = 0.5)')
    axes[1].set_title(f'Curva ROC\n{model_name}')
    axes[1].legend(loc='lower right')
    
    plt.tight_layout()
    plt.show()

df_anomaly_detection_summary = pd.DataFrame(lista_result)
```
